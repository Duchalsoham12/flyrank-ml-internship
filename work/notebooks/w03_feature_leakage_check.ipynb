```python
import pandas as pd
import numpy as np

file_path = "/content/flyrank-starter/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(file_path)

print("Data loaded successfully")
print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
numeric_features = [
    "search_volume",
    "competition",
    "cpc",
    "word_count",
    "char_count",
    "ctr",
    "avg_position",
    "engagement_rate",
    "scroll_rate",
    "ai_traffic_pct"
]

categorical_features = [
    "competition_level",
    "content_type",
    "main_intent"
]

X = df[numeric_features + categorical_features].copy()

# Fill missing numeric values
for col in numeric_features:
    X[col] = X[col].fillna(X[col].median())

# Fill missing categorical values
for col in categorical_features:
    X[col] = X[col].fillna("missing")

# Convert categorical columns to numbers
X = pd.get_dummies(
    X,
    columns=categorical_features
)

print("Feature vector created")
print("Rows:", len(X))
print("Features:", len(X.columns))
print("Missing values:", X.isna().sum().sum())

display(X.head())
```

## 3. The leakage hunt

*Attack your own features: label-derived columns, future windows, product flags. Show the test.*

```python
# ML-05 Section 3: Leakage Hunt

target_column = "trend_direction"

# Fields that could leak the outcome
leakage_candidates = [
    "trend_direction",
    "trend_pct",
    "position_tier",
    "impression_tier"
]

print("LEAKAGE CHECK")
print("=" * 40)

for col in leakage_candidates:
    print(f"{col}: {'FOUND' if col in X.columns else 'NOT USED'}")

print("\nTarget included in feature vector:",
      target_column in X.columns)

# Check that none of the leakage candidates are in X
leakage_in_features = [
    col for col in leakage_candidates
    if col in X.columns
]

print("Leakage fields in X:", leakage_in_features)

if len(leakage_in_features) == 0:
    print("\nPASS: No identified leakage fields are being used.")
else:
    print("\nWARNING: Review the fields listed above.")
```

## 4. What I excluded and why

*The list of fields you refused to use — with one line of why each.*

```python
# ML-05 Section 4: Excluded fields and reasons

excluded_features = {
    "content_id": "Identifier only; not a meaningful predictive feature.",
    "client_id": "Identifier only; may cause the model to memorize client-specific patterns.",
    "trend_direction": "Target label; must not be used as an input feature.",
    "trend_pct": "Directly describes the observed trend outcome and may cause target leakage.",
    "position_tier": "Derived ranking category; excluded to reduce overlap with the performance outcome.",
    "impression_tier": "Derived impression category; excluded to reduce overlap with the performance outcome."
}

print("EXCLUDED FEATURES")
print("=" * 40)

for feature, reason in excluded_features.items():
    print(f"\n{feature}")
    print("Reason:", reason)

print("\nExcluded fields:", len(excluded_features))
```

I excluded identifiers because they do not represent useful content characteristics. I excluded the target and direct trend fields because they could leak the outcome into the model. Derived ranking and impression tiers were also excluded to reduce overlap with the target.

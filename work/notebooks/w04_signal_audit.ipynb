## 2. Signal test #1 / #2 / #3 (verdict each)

*Three safe signals, each with a mini-test and a verdict: CONFIRMED / OPPOSITE / MIXED / FALSE.*

```python
# Check the trend categories
print("Trend direction values:")
print(df["trend_direction"].value_counts(dropna=False))
```

```python
# Signal 1: days since last update
print("SIGNAL 1: Freshness")

display(
    df.groupby("trend_direction")["days_since_last_update"]
      .agg(["count", "mean", "median"])
      .round(2)
)
```

```python
# Signal 2: Engagement rate
print("SIGNAL 2: Engagement")

display(
    df.groupby("trend_direction")["engagement_rate"]
      .agg(["count", "mean", "median"])
      .round(4)
)
```

```python
# Signal 3: AI traffic
print("SIGNAL 3: AI Traffic")

display(
    df.groupby("trend_direction")["ai_traffic_pct"]
      .agg(["count", "mean", "median"])
      .round(4)
)
```

## 3. The flag-linked test

*Pick a signal one of FlyRank's real flags relies on. Does the data support the rule's assumption?*

```python
# Section 3: Flag-linked test
# Test whether freshness_tier reflects days_since_last_update

print("Freshness tiers:")
print(df["freshness_tier"].value_counts(dropna=False))

print("\nDays since last update by freshness tier:")

freshness_test = (
    df.groupby("freshness_tier")["days_since_last_update"]
      .agg(
          count="count",
          mean="mean",
          median="median",
          min="min",
          max="max"
      )
      .sort_values("median")
      .round(2)
)

display(freshness_test)
```

```python
# Check whether the freshness tiers are ordered consistently
tier_check = (
    df.groupby("freshness_tier")["days_since_last_update"]
      .median()
      .sort_values()
)

print("Median days since update by tier:")
display(tier_check)

print("\nNumber of unique freshness tiers:", df["freshness_tier"].nunique())
```

tested whether the freshness flag is supported by its underlying signal, days_since_last_update. If the freshness tiers show clearly different measured update ages, the rule's assumption is supported; if the values overlap heavily, the flag should be treated cautiously.
